```python
import sys
import os
from dotenv import load_dotenv

# The old API version
from openai import AzureOpenAI

# The new API version
from openai import OpenAI

%load_ext autoreload
%autoreload 2
import agentix

print(f'Package version: {agentix.__version__}')
print(f'Authors:         {agentix.__authors__}')
print(f'Python version:  {sys.version}')
```

```python
# Load environmental variables
load_dotenv()
data_root = os.environ.get('AZURE_API_KEY')
if data_root is None:
    print('Edit the "env"file and copy it to the root of the repository.') 
```

```python
api_key = os.environ['AZURE_API_KEY']
azure_endpoint = os.environ['AZURE_API_ENDPOINT']
print(f'AZURE ENDPOINT LOADED: {azure_endpoint}')
model = 'gpt-5'
```

### The OLD Azure OpenAI version ###
This usage is described in the Azure AI L4 Platform site:

https://hu.sharepoint.com/sites/AzureAI

```python
# Prompt/User input
message = [
  {"role": "system", "content": "You are a helpful AI assistant."}, 
  {"role": "user", "content": "Nice to meet you."}
]

# Build your client
client = AzureOpenAI(azure_endpoint=azure_endpoint, api_key=api_key, api_version='2024-10-21')

# Submit your request
response = client.chat.completions.create(
  model=model, # Model deployment id
  messages=message,        
  temperature=1.0,
  max_completion_tokens=250,
)

print(response.choices[0].message.content)
```

### Azure OpenAI in Microsoft Foundry Models v1 API ###
The v1 API simplifies authentication, removes the need for dated api-version parameters, and supports cross-provider model calls.

```python
base_url = f'{azure_endpoint}/openai/v1'

message = [
    {"role": "system", "content": "You are a helpful AI assistant."},
    {"role": "user", "content": "Write a one-sentence poem about unicorns"}
]

client = OpenAI(base_url=base_url, api_key=api_key, default_headers={'api-key': api_key})

response = client.chat.completions.create(model=model,
                                          messages=message,
                                          temperature=1.0,
                                          max_completion_tokens=500)
print('Response')
print()
print(response.choices[0].message.content)
print()
print(f'finish_reason: {response.choices[0].finish_reason}')   # likely "length"
print()
usage = response.usage
print(usage.model_dump_json(indent=2))
```

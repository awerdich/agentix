```python
# Prompt/User input
message = [
  {"role": "system", "content": "You are a helpful AI assistant."}, 
  {"role": "user", "content": "Nice to meet you."}
]

# Build your client
client = AzureOpenAI(azure_endpoint=azure_endpoint, api_key=api_key, api_version='2024-10-21')

# Submit your request
response = client.chat.completions.create(
  model=model, # Model deployment id
  messages = message,        
  temperature=0.5,
  max_tokens=250,
)

print(response.choices[0].message.content)
```

### Azure OpenAI in Microsoft Foundry Models v1 API ###
The v1 API simplifies authentication, removes the need for dated api-version parameters, and supports cross-provider model calls.

```python
base_url = f'{azure_endpoint}/openai/v1'

message = [
    {"role": "system", "content": "You are a helpful AI assistant."},
    {"role": "user", "content": "Can you explain why the sky is blue?"}
]

client = OpenAI(base_url=base_url, api_key=api_key, default_headers={'api-key': api_key})

response = client.chat.completions.create(model=model,
                                          messages=message,
                                          temperature=1.0,
                                          max_completion_tokens=500)
print('Response')
print()
print(response.choices[0].message.content)
print()
print(f'finish_reason: {response.choices[0].finish_reason}')   # likely "length"
print()
usage = response.usage
print(usage.model_dump_json(indent=2))
```
